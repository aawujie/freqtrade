## 1. 下载数据

- --erase：重下，覆盖式（先删再下）。
- --prepend：在已有数据前面补更早历史，保留现有数据。

```python
!docker compose run --rm freqtrade download-data \
  --config /freqtrade/user_data/config_composed.json \
  --trading-mode futures \
  --timeframe 5m \
  --data-format-ohlcv feather \
  --pairs ASTER/USDT:USDT \
  --timerange 20250901-20250923 \
  --erase
```

```python
!docker compose run --rm freqtrade download-data \
  --config /freqtrade/user_data/config_composed.json \
  --trading-mode spot \
  --timeframe 5m \
  --data-format-ohlcv feather \
  --pairs ETH/USDT BTC/USDT \
  --prepend \
  --timerange 20240101-20241231
```
